```python
import kegg_pull.rest as r
import kegg_pull.pull as p
import kegg_pull.entry_ids as ei
import kegg_pull.map as kmap
import pandas as pd
from pathlib import Path
```

```python
repo_folder = Path('../../..')
data_folder  = repo_folder / 'data'/ 'this_project' / '6_transporterKO' / 'Z_div'
```

```python
single_pull = p.SinglePull()
```

```python
# Get all KOs from E. coli
temp = kmap.entries_link(entry_ids=['T00007'], target_database='ko')
KOs = []
for key, value in temp.items():
    KOs+=list(value)
KOs = list(set(KOs))
```

```python
i0 = 0
ort_to_rxn = {}
for i in range(100, 3200, 100):
    print(i)
    temp = kmap.entries_link(entry_ids=KOs[i0:i], target_database='reaction')
    ort_to_rxn.update(temp)
    i0 = i
```

```python
reactions = []
for key, value in ort_to_rxn.items():
    reactions+=list(value)
reactions  = list(set(reactions))
```

```python
rxn_to_cmp = {}
i0 = 0
for i in range(100, 1700, 100):
    print(i)
    temp = kmap.entries_link(entry_ids=reactions[i0:i], target_database='compound')
    rxn_to_cmp.update(temp)
    i0 = i
```

```python
compounds = []
for key, value in rxn_to_cmp.items():
    compounds+=list(value)
compounds  = list(set(compounds))
```

```python
kegg_compounds_list = [x.split(':')[-1] for x in compounds]
```

```python
pd.Series(kegg_compounds_list).to_csv(data_folder / 'A_ecoli_k12_kegg_ids_from_kegg.csv')
```
