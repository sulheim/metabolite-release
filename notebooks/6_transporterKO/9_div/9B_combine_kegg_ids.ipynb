# Combine KEGG IDs
Collected data from BioCyC, KEGG and the genome-scale metabolic model iML1515.

```python
repo_folder = Path('../../..')
data_folder  = repo_folder / 'data'/ 'this_project' / '6_transporterKO' / 'Z_div'
```

```python
kegg_ecoli_kegg_ids = pd.read_csv(data_folder / 'A_ecoli_k12_kegg_ids_from_kegg.csv', index_col=0)
iML1515_ecoli_kegg_df = pd.read_csv(data_folder / 'B_ecoli_iML1515_kegg_ids.csv')

# biocyc_ecoli_kegg_ids = pd.read_csv('../data/biocyc_all_compounds_Escherichia-coli-K-12-substr.-MG1655_kegg_only.txt', sep = '\t')
# ecoli_kegg_ids = sorted(list(set(ecoli_kegg_df['KEGG ID'].values)))
```
