```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_folder = Path('../../../data')
fn_selected_transporters = data_folder / 'this_project' / '6_transporterKO' / 'B_selected_transporters_with_expression_levels.csv'
df = pd.read_csv(fn_selected_transporters, index_col=0)
```

```python
df.tail()
```

```python
values = df.loc[:, 'Location'].value_counts()
values.plot(kind='pie', cmap = 'tab20', autopct = lambda x: '{:.0f}'.format(x*values.sum()/100))
```

```python
df
```

```python
values = df.loc[:, 'Metabolite class'].value_counts()
values.plot(kind='pie', cmap = 'tab20', autopct = lambda x: '{:.0f}'.format(x*values.sum()/100))
```

```python
values = df.loc[:, 'Class'].value_counts()
values.plot(kind='pie', cmap = 'tab20', autopct = lambda x: '{:.0f}'.format(x*values.sum()/100), fontsize = 8)
```

```python
values = df.loc[:, 'Mechanism/specific type'].value_counts()
values.plot(kind='pie', cmap = 'tab20', autopct = lambda x: '{:.0f}'.format(x*values.sum()/100))
```

```python
values = df.loc[:, 'Type'].value_counts()
ax = values.plot(kind='pie', cmap = 'tab20', autopct = lambda x: '{:.0f}'.format(x*values.sum()/100), title='')
ax.set_ylabel('')
```

```python
values = df.loc[:, 'Expected direction'].value_counts()
ax = values.plot(kind='pie', cmap = 'tab20', autopct = lambda x: '{:.0f}'.format(x*values.sum()/100)) 
ax.set_ylabel('')
```
